```python
Customer_Engagement['Frequency_level']=pd.cut(
    Customer_Engagement['Purchase_Frequency'],
    bins=[0,1,5,float('inf')],
    labels=['low','Medium','High']
)
```

```python
Customer_Engagement.head()
```

```python
Customer_Engagement['Recency_level']=pd.cut(
    Customer_Engagement['Recency_Days'],
    bins=[0,18,142,float('inf')],
    labels=['Recent','Medium','Not Recent']
)
```

```python
Customer_Engagement.head()
```

```python
def get_engagement_level(row):
  if row['Frequency_level']=='High' and row['Recency_level']=='Recent':
    return 'High'
  elif row['Frequency_level']=='Medium' and row['Recency_level']=='Recent':
    return 'High'
  elif row['Frequency_level']=='High' and row['Recency_level']=='Medium':
    return 'High'
  elif row['Recency_level']=='Not Recent':
    return 'Low'
  else:
    return 'Medium'
```

```python
Customer_Engagement['Engagement_level']=Customer_Engagement.apply(
    get_engagement_level,
    axis=1
)
```

```python
Monetary=df.groupby('CustomerID')['Total Amount'].sum()
```

```python
rfm=pd.DataFrame({
    'Recency_Days':recency_days,
    'Frequency_Days':Purchase_Frequency,
    'Monetary':Monetary
})
```

```python
rfm.head()
```

```python
rfm['R_Score']=pd.qcut(
    rfm['Recency_Days'],
    4,
    labels=[4,3,2,1]
)
```

```python
rfm.head()
```

```python
rfm['F_Score']=pd.qcut(
    rfm['Frequency_Days'].rank(method='first'),
    4,
    labels=[1,2,3,4]
)
```

```python
rfm.head()
```

```python
rfm['M_Score']=pd.qcut(
    rfm['Monetary'],
    4,
    labels=[1,2,3,4]
)
```

```python
rfm.head()
```

```python
rfm['RFM_Score']=(
    rfm['R_Score'].astype(str)+
    rfm['F_Score'].astype(str)+
    rfm['M_Score'].astype(str)
)
```

```python
rfm.head()
```

```python
rfm['RFM_Score'].value_counts().head(10)
```

```python
def get_segment(row):
  if row['R_Score']==4 and row['F_Score']==4 and row['M_Score']==4:
    return'Champions'
  elif row['R_Score']>=3 and row['F_Score']>=3:
    return'Loyal Customers'
  elif row['M_Score']==4:
    return'Big Spenders'
  elif row['R_Score']>=3 and row['F_Score']<=2:
    return'New Customers'
  elif row['R_Score']<=2 and row['F_Score']>=3:
    return'At Risk'
  else:
    return'Low Engagement'
```

```python
rfm['Segment']=rfm.apply(get_segment,axis=1)
```
